```python
from langchain_ollama import OllamaLLM

llm=OllamaLLM(
    model="llama3:latest",
    temperature=0,
    top_k=3,
    num_ctx=2048,
    num_predict=500,
    verbose=False
)
```

```python
llm.invoke("what is 33-33+12?")
```

```python
from langchain.prompts import PromptTemplate

template="""<s><SYS>you are a personal assistant[/SYS][INST]
this the format:{user_input}
[/INST]"""
prompt= PromptTemplate(template=template, input_variables=["user_input"])
```

```python
basic_chain=prompt | llm
```

```python
basic_chain.invoke({"user_input":"what?"})
```

```python
def get_lazy(query):
    return basic_chain.invoke(query)

#this worked because llama is not that strict with format nhi to .invoke({"user_input":"query"})
```

```python
get_lazy("what is this")
```
